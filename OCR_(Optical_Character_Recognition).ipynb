```python
import easyocr

reader = easyocr.Reader(['bn'])

result = reader.readtext("/content/images (2).png")

for detection in result:
    print("Detected Text:", detection[1])
    # print("Confidence:", detection[2])
```

# PaddleOCR

```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    use_angle_cls=True,
    lang='en'
)

result = ocr.ocr("images.png", cls=True)

for line in result:
    for word in line:
        print("Text:", word[1][0])
        # print("Confidence:", word[1][1])
```
